```python
def sort012(arr):
    arr.sort()

if __name__ == "__main__":
    arr = [0, 1, 2, 0, 1, 2]
    sort012(arr)
    
    for num in arr:
        print(num, end=" ")
```

```python
class Solution:

    def firstSearch(self, arr, k):
        left, right = 0, len(arr) - 1
        result = -1

        while left <= right:
            mid = left + (right - left) // 2

            if arr[mid] == k:
                result = mid
                right = mid - 1  # Search in the left half for the first occurrence
            elif arr[mid] < k:
                left = mid + 1
            else:
                right = mid - 1

        return result
```
