```python
class Vehicle:
    def __init__(self, vehicle_type):
        self.vehicle_type = vehicle_type


class Automobile(Vehicle):
    def __init__(self, vehicle_type, year, make, model, door, roof):
        Vehicle.__init__(self, vehicle_type)
        self.year = year
        self.make = make
        self.model = model
        self.doors = door
        self.roof = roof

vehicle_type = input("Enter the type of car you have ")
year = input("Enter the year of your car ")
make = input("Enter the make of your car ")
model = input("Enter the model of your car ")
door = input("Enter the number of door (2 or 4) your car has ")
roof = input("Enter if your roof is solid or a sun roof ")


car = Automobile(vehicle_type, year, make, model, door, roof)

print("Car Information")
print("Car type: ", car.vehicle_type)
print("Year: ", car.year)
print("Make: ", car.make)
print("Model: ", car.model)
print("Number of doors: ", car.door)
print("Type of roof: ", car.roof)
```
